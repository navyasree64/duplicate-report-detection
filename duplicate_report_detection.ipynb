```python
import json,pandas as pd
from google.colab import files
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
with open('/content/Duplicate_Reports.json','r') as file:
  data=json.load(file)
reports=data['hits']['hits']
df=pd.DataFrame([{"id":r["_source"]["id"],
                  "text":r["_source"]["text"]}
                 for r in reports])
print("Total reports:",len(df))
print(df)
df["text"]=df["text"].fillna("").astype(str)
df=df[df["text"].str.strip()!=""].reset_index(drop=True)
vectorizer=TfidfVectorizer(ngram_range=(1,2))
vectors=vectorizer.fit_transform(df["text"])
similarity=cosine_similarity(vectors)
threshold=0.4
for i in range(len(df)):
  for j in range(i+1,len(df)):
    score=similarity[i][j]
    if score>=threshold:
      print(
          df.iloc[i]['id'],'and',df.iloc[j]['id'],
          "Similarity:",round(score,3))
r2_index=df.index[df["id"]=="R2"][0]
r3_index=df.index[df["id"]=="R3"][0]
print("R2 and R3 similarity:",round(similarity[r2_index,r3_index],3))
```
